```python
def reflex_vacuum_agent(location,status):
  if status == "Dirty":
    return "Suck"
  elif location == "A":
    return "Right"
  elif location == "B":
    return "Left"
```

```python
def move_towards_goal(position,goal):
  while position < goal:
    position +=1
    print(f"Current Position: {position}")
  print("Reached the goal!")
```
